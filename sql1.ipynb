### 1.Select
It is used for selecting or retriving the data as you needed

```python
cur.execute("select * from employee_demographics")
result = cur.fetchall()
for row in result:
    print(row)
```

### 2. AND OR NOT

```python
cur.execute("select * from employee_demographics where age>40 and gender = 'male'")
result = cur.fetchall()
for row in result:
    print(row)
```

```python
cur.execute("select * from employee_demographics where age>40 or gender = 'male'")
result = cur.fetchall()
for row in result:
    print(row)
```

```python
cur.execute("select * from employee_demographics where age>40 and not gender = 'male'")
result = cur.fetchall()
for row in result:
    print(row)
```

### Like Operator

% and _

```python
cur.execute("select * from employee_demographics where first_name like 'A%'")
result = cur.fetchall()
for row in result:
    print(row)
```

```python
cur.execute("select * from employee_demographics where first_name like '%a'")
result = cur.fetchall()
for row in result:
    print(row)
```
